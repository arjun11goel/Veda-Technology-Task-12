```python
import numpy as np

# Create two NumPy arrays using np.array()
array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([2, 4, 5, 8, 10])

print("Array 1:", array1)
print("Array 2:", array2)
print("Shape of Array 1:", array1.shape)
print("Shape of Array 2:", array2.shape)
print("-" * 30)

# Perform vectorized mathematical operations
addition_result = array1 + array2
subtraction_result = array1 - array2
multiplication_result = array1 * array2
division_result = array1 / array2

# Display results
print("Addition (Array 1 + Array 2):", addition_result)
print("Subtraction (Array 1 - Array 2):", subtraction_result)
print("Multiplication (Array 1 * Array 2):", multiplication_result)
print("Division (Array 1 / Array 2):", division_result)
```

### Short Explanation of Vectorized Operations
Vectorized operations in NumPy allow you to apply mathematical operations to entire arrays at once, rather than iterating through elements one by one using loops. When we execute `array1 + array2`, NumPy internally handles the element-by-element addition at the C-language level, which is much more efficient and requires less code than writing a Python `for` loop.

### Interview Questions

**1. What is vectorization?**
Vectorization refers to the process of executing operations on entire arrays or batches of data simultaneously, rather than processing individual elements sequentially using loops. In libraries like NumPy, vectorized operations are delegated to optimized, pre-compiled C code, leading to significant performance improvements and cleaner syntax.

**2. What happens when two NumPy arrays are added?**
When two NumPy arrays of the same shape are added (e.g., `array1 + array2`), NumPy performs element-wise addition. The first element of the first array is added to the first element of the second array, the second to the second, and so on, resulting in a new array containing the sums. If the arrays have different shapes, NumPy may attempt to "broadcast" them to a common shape if the dimensions are compatible.

**3. Why is NumPy faster than ordinary Python lists for numerical operations?**
NumPy is faster for several reasons:
*   **Contiguous Memory:** NumPy arrays store elements of the same data type in contiguous memory blocks, allowing for faster access and better cache utilization compared to Python lists, which store pointers to scattered objects.
*   **Pre-compiled C Code:** NumPy operations are implemented in highly optimized C code, avoiding the overhead of Python's interpreter and dynamic type checking during loops.
*   **Vectorization:** NumPy leverages vectorization to perform operations on entire data blocks simultaneously, whereas standard Python lists require explicit, slow iteration.
